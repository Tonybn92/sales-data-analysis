```python
super_store['Segment'].unique() ## Unique segments in the dataset
# Son 3 segmentos: 'Consumer', 'Corporate', 'Home Office'
```

```python
numberPerSegment = super_store['Segment'].value_counts(normalize=True) * 100  ## Count of each segment in the dataset
segmentPercentages = round(numberPerSegment, 2) ## Round the percentages to 2 decimal places
print(segmentPercentages.astype(str) + " %")
# Consumer is the most common segment, followed by Corporate and Home Office.
```

```python
"""The Consumer segment represents approximately 52% of all transactions, making it the dominant customer group.

This indicates that the business relies heavily on individual consumers rather than corporate clients."""
```

```python
super_store['Ship Mode'].unique() ## Unique ship modes in the dataset
# Son 4 ship modes: 'First Class', 'Same Day', 'Second Class', 'Standard Class'
```

```python
super_store['Ship Mode'].value_counts() ## Count of each ship mode in the dataset
# Standard Class is the most common ship mode, followed by Second Class, First Class, and Same Day.
```

```python
"""Standard Class is the most frequently used shipping method, accounting for the majority of orders.

This suggests that customers prefer cost-effective shipping options over faster delivery methods."""
```

## Key Insights from Categorical Analysis

- The business operates across three main product categories.
- The West region is the most active market.
- The Consumer segment dominates, indicating reliance on individual customers.
- Standard Class is the preferred shipping method, suggesting cost sensitivity among customers.

## NUMERICAL ANALYSIS (INSIGHTS)
- The average sales value is approximately 229.85, indicating that most transactions are relatively small.

However, the dataset contains extreme values, with sales reaching up to 22,638, suggesting the presence of high-value transactions.

- The average profit is around 28.65, but there are significant negative values, with losses as low as -6,599.

This indicates that some transactions result in substantial losses, which may be driven by high discounts or unprofitable products.

- There are 1,871 transactions with negative profit, showing that a considerable portion of sales are not profitable.

- The maximum discount is 80%, which is very high and could be a key factor contributing to negative profits.

```python
(super_store['Sales'][super_store['Discount'] == 0.8].sum()) ## Total sales for transactions with 80% discount
totalSales = super_store['Sales'].sum() ## Total sales for all transactions
percentageSales = (super_store['Sales'][super_store['Discount'] == 0.8].sum() / totalSales) * 100 ## Percentage of sales from transactions with 80% discount
print(f"The percentage of sales from transactions with an 80% discount is: {percentageSales:.2f} %")
```

## Final Analysis INSIGHT NUMERICAL

### High Discount Impact Analysis

Transactions with an 80% discount represent only 0.74% of total sales and account for 300 transactions.

Despite their low contribution to total revenue, these transactions generate a total loss of -30,539.

This indicates that high discount levels are significantly impacting profitability and may not be a sustainable pricing strategy.

A small subset of highly discounted transactions is responsible for a disproportionately large share of total losses.

### Distribution Insights

The dataset shows a highly skewed distribution in both Sales and Profit.

Most transactions are low-value and generate small profits, while a small number of transactions account for very high values.

This indicates that the business relies on many small transactions, with occasional high-value sales contributing disproportionately to total revenue.
